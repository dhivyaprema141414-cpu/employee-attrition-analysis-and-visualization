# Step 3: Exploratory Data Analysis & Visualization

## Objective

The objective of this section is to analyze employee attrition patterns through visualizations and identify the factors contributing to employee turnover.

The insights generated from these visualizations will help HR teams make data-driven decisions and improve employee retention strategies.

## 3.1 Employee Attrition Overview

### Business Problem

Many employees are leaving the company, but management needs to understand the overall attrition situation.

### Objective

Analyze the distribution of employees who stayed and those who left the company.

```python
plt.figure(figsize=(12,8))

ax = sns.countplot(
    x='Attrition',
    data=df,
    palette={
    'No': '#3A86FF',   # Royal Blue
    'Yes': '#C1121F'  # Dark Crimson Red
}
)

for p in ax.patches:
    ax.annotate(
        f'{int(p.get_height())}',
        (p.get_x() + p.get_width()/2, p.get_height()),
        ha='center',
        va='bottom'
    )

plt.title("Employee Attrition Distribution", fontsize=14, fontweight='bold')
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

plt.show()
```

### Insight

This visualization shows the number of employees who stayed and left the company.

A significant number of employees have left the organization, indicating that employee attrition is an important business issue that requires further investigation.

## 3.2 Department-wise Attrition Analysis

### Business Problem

Some departments may experience higher employee turnover than others.

### Objective

Compare attrition rates across departments to identify departments with higher employee turnover.

```python
plt.figure(figsize=(12,8))

ax = sns.countplot(
    x='Department',
    hue='Attrition',
    data=df,
    palette={
    'No': '#3A86FF',   # Royal Blue
    'Yes': '#C1121F'  # Dark Crimson Red
}
)

for container in ax.containers:
    ax.bar_label(container)
sns.set_style("whitegrid")
plt.title("Attrition by Department", fontsize=14, fontweight='bold')
plt.xlabel("Department", fontsize=12)
plt.ylabel("Number of Employees", fontsize=12)

plt.legend(title="Attrition")
plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```

### Insight

This visualization helps identify departments with higher attrition levels.

Departments showing higher attrition may require focused retention strategies and further investigation.

## 3.3 Attrition by Job Role

### Business Problem

Certain job roles may have higher turnover rates.

### Objective

Identify job roles that experience the highest employee attrition.

```python
plt.figure(figsize=(15,15))

ax = sns.countplot(
    y='JobRole',
    hue='Attrition',
    data=df,
    palette={'No':'#3A86FF', 'Yes':'#C1121F'}
)

for container in ax.containers:
    ax.bar_label(container, fontsize=18, fontweight='bold')

plt.title("Attrition by Job Role", fontsize=28, fontweight='bold')
plt.xlabel("Number of Employees", fontsize=24, fontweight='bold')
plt.ylabel("Job Role", fontsize=24, fontweight='bold')

plt.xticks(fontsize=22, fontweight='bold')
plt.yticks(fontsize=22, fontweight='bold')

plt.legend(
    title='Attrition',
    fontsize=20,          # Legend text
    title_fontsize=22     # Legend title
)

plt.grid(axis='x', linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```

### Insight

This chart highlights the job roles with the highest employee turnover.

HR teams can use this information to develop targeted retention strategies for specific roles.

## 3.4 Overtime vs Attrition

### Business Problem

Excessive overtime may contribute to employee resignations.

### Objective

Analyze the relationship between overtime and employee attrition.

```python
plt.figure(figsize=(12,8))

ax = sns.countplot(
    x='OverTime',
    hue='Attrition',
    data=df,
    palette={'No':'#3A86FF', 'Yes':'#C1121F'}
)

for c in ax.containers:
    ax.bar_label(c)

plt.title("Overtime vs Attrition", fontsize=18, fontweight='bold')
plt.xlabel("OverTime")
plt.ylabel("Number of Employees")
plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.show()
```

### Insight

Employees who work overtime may show higher attrition rates compared to employees who do not work overtime.

This suggests that work-life balance could be an important factor influencing employee retention.

## 3.5 Salary Impact on Attrition

### Business Problem

Employees may leave the company due to low salary.

### Objective

Analyze the relationship between monthly income and employee attrition.

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    x='Attrition',
    y='MonthlyIncome',
    data=df,
    palette={'No':'#3A86FF', 'Yes':'#C1121F'}
)

plt.title("Monthly Income vs Attrition", fontsize=18, fontweight='bold')
plt.xlabel("Attrition")
plt.ylabel("Monthly Income")

plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.show()
```

### Insight

Employees who left the company generally have lower monthly income compared to employees who stayed.

This suggests that compensation may be an important factor influencing employee retention.

## 3.6 Job Satisfaction Analysis

### Business Problem

Low job satisfaction may increase employee turnover.

### Objective

Analyze how job satisfaction levels affect employee attrition.

```python
plt.figure(figsize=(12,8))

ax = sns.countplot(
    x='JobSatisfaction',
    hue='Attrition',
    data=df,
    palette={'No':'#3A86FF', 'Yes':'#C1121F'}
)

for c in ax.containers:
    ax.bar_label(c)

plt.title("Job Satisfaction vs Attrition", fontsize=18, fontweight='bold')
plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.show()
```

### Insight

Employees with lower job satisfaction levels show higher attrition rates.

Improving employee satisfaction may help reduce turnover.

## 3.7 Age Analysis

### Business Problem

Employee age may influence resignation patterns.

### Objective

Analyze attrition across different age groups to identify which age ranges are more likely to leave the company.

```python
plt.figure(figsize=(10,6))

sns.histplot(
    data=df,
    x='Age',
    hue='Attrition',
    kde=True,
    palette={'No':'#3A86FF', 'Yes':'#C1121F'}
)

plt.title("Age Distribution by Attrition", fontsize=18, fontweight='bold')
plt.xlabel("Age")
plt.ylabel("Count")

plt.show()
```

### Insight

This visualization helps identify whether younger or older employees are more likely to leave the company.

## 3.8 Experience and Retention

### Business Problem

New employees may leave before becoming long-term contributors.

### Objective

Analyze the relationship between years at company and attrition.

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    x='Attrition',
    y='YearsAtCompany',
    data=df,
    palette={'No':'#3A86FF', 'Yes':'#C1121F'}
)

plt.title("Years at Company vs Attrition", fontsize=18, fontweight='bold')
plt.xlabel("Attrition")
plt.ylabel("Years at Company")

plt.show()
```

### Insight

This analysis helps determine whether employees are leaving during their early years or after long periods with the company.

## 3.9 Correlation Analysis

### Business Problem

Multiple factors may influence employee attrition simultaneously.

### Objective

Identify relationships between important numerical features.

```python
plt.figure(figsize=(14,10))

corr = df.corr(numeric_only=True)

sns.heatmap(
    corr,
    cmap='coolwarm',
    annot=True,
    fmt='.1f'
)

plt.title("Correlation Heatmap", fontsize=18, fontweight='bold')

plt.show()
```

### Insight

The heatmap reveals relationships among age, income, experience, satisfaction, and other employee-related factors.

Strong correlations may indicate variables that influence employee retention and turnover.
